## 🌍 Contexte et introduction  

Avec l’évolution des technologies ferroviaires, les trains modernes ne sont plus de simples moyens de transport mécaniques : 
ce sont de véritables **systèmes cyber-physiques complexes**, dotés de capteurs avancés, de calculateurs embarqués et de réseaux de communication à haut débit.  

La SNCF et ses partenaires industriels (Alstom, Siemens, Hitachi…) travaillent actuellement sur le projet **Train Autonome**.  
Un TGV autonome doit être capable de :  

- **Percevoir son environnement** (obstacles, signaux, conditions extérieures) grâce à des capteurs variés :  
  **LIDAR, caméras, radars, GPS/IMU, capteurs thermiques**, mais aussi des capteurs classiques (tachymètre, balises ETCS, sondes de pression et température).  

- **Traiter l’information en temps réel** grâce à un calculateur central (**TCMS – Train Control and Management System**) et à des modules d’**IA embarquée** pour la fusion de données.  

- **Communiquer** avec l’ensemble des sous-systèmes du train via des réseaux internes (**MVB, Ethernet train**) et avec le sol via des réseaux externes (**GSM-R aujourd’hui, FRMCS/5G demain**).  

- **Restituer l’information** au conducteur (s’il existe encore un rôle de supervision humaine), au système de supervision embarqué, et au centre de contrôle à distance via des **plateformes cloud ferroviaires**.  

---

Autrement dit, le TGV autonome fonctionne comme un **organisme vivant** :  
- ses **capteurs** jouent le rôle des organes sensoriels (vue, ouïe, toucher),  
- ses **réseaux** sont le système nerveux,  
- son **calculateur central (TCMS + IA)** agit comme le cerveau,  
- et son **IHM + cloud** jouent le rôle du langage et de la communication externe.

## 📘 Partie 2 – Capteurs modernes du TGV autonome  

### 1. LIDAR (Light Detection and Ranging)  
- **Principe** : émission d’un faisceau laser → mesure du temps de retour → reconstruction 3D de l’environnement.  
- **Rôle** : détection d’obstacles, aide à la navigation.  

### 2. Caméras embarquées  
- **Principe** : vision artificielle, analyse d’images par IA.  
- **Rôle** : lecture signaux, détection obstacles, surveillance.  

### 3. GPS + Inertial Measurement Unit (IMU)  
- **Principe** : GPS (position absolue) + IMU (gyroscope + accéléromètre).  
- **Rôle** : localisation continue et précise, complément ETCS/KVB.  

### 4. Capteurs environnementaux avancés  
- **Radar Doppler** → vitesse instantanée.  
- **Capteurs thermiques/IR** → détection chaleur, sécurité.  
- **Capteurs acoustiques** → analyse bruit (roulements, freins).

### Tableau 1 – Étude des capteurs modernes  

| Capteur | Principe | Grandeur mesurée | Avantage | Limite | Exemple concret |
|---------|----------|------------------|----------|--------|-----------------|
| LIDAR | Émission d'un faisceau laser, mesure du temps de retour, reconstruction 3D de l'environnement | Distance / position 3D des obstacles | Détection précise en 3D, fonctionne de nuit | Coût élevé, sensible aux intempéries (pluie, brouillard) | Détection d'un obstacle sur la voie |

## 📘 Partie 3 – Réseaux ferroviaires  


### 🔹 Réseaux internes (dans le train)
- **MVB (Multifunction Vehicle Bus)**  
  - Bus série normalisé (IEC 61375).  
  - Fiable, robuste, faible débit (~1,5 Mbit/s).  
  - Utilisé pour les systèmes critiques : freins, portes, traction.  

- **Ethernet Train (ETB / ETBN)**  
  - Adaptation de l’Ethernet classique.  
  - Haut débit (100 Mbit/s – 1 Gbit/s).  
  - Supporte vidéosurveillance, Wi-Fi passagers, diagnostics.  

### 🔹 Réseaux sol-train (communication avec l’infrastructure)
- **GSM-R (Global System for Mobile – Railway)**  
  - Basé sur la 2G (GSM).  
  - Communication conducteur ↔ régulateur.  
  - Débit limité (~9,6 kbit/s).  
  - Fin de vie prévue vers 2030.  

- **FRMCS (Future Railway Mobile Communication System)**  
  - Basé sur la 5G ferroviaire.  
  - Très haut débit, faible latence.  
  - Support IoT ferroviaire et conduite autonome.  
  - Remplacera GSM-R dans les prochaines années.  

### 🔹 Supervision distante (cloud ferroviaire)
- Transmission des données en temps réel au **centre de contrôle**.  
- Utilisation de réseaux IP sécurisés (fibre, 5G, VPN).  
- Sert à la maintenance prédictive et à la surveillance globale.  



### -Schéma simplifié de la communication

```
Capteurs (vitesse, portes, température, signalisation)
        ↓
Réseaux internes (MVB / Ethernet train)
        ↓
Calculateur central (TCMS)
        ↓
Réseaux sol-train (GSM-R / FRMCS)
        ↓
Centre de contrôle & Supervision distante
```


| Réseau | Domaine | Débit | Avantages | Limites | Exemple |
|--------|---------|-------|-----------|---------|---------|
| MVB | Interne train | ~1,5 Mbit/s | Fiable, robuste, normalisé (IEC 61375) | Faible débit, peu adapté aux applications lourdes | Communication freins, portes, traction |
| Ethernet Train | Interne train | 100 Mbit/s – 1 Gbit/s | Haut débit, supporte plusieurs applications simultanées | Plus coûteux et complexe à sécuriser que le MVB | Vidéosurveillance, Wi-Fi passagers, diagnostics |
| GSM-R | Sol-train | ~9,6 kbit/s | Standard éprouvé, couverture large sur le réseau ferré | Débit très limité, technologie 2G en fin de vie (~2030) | Communication conducteur ↔ régulateur |
| FRMCS | Sol-train | Très haut débit (5G) | Faible latence, support IoT et conduite autonome | Déploiement encore en cours, nécessite nouvelle infrastructure | Transmission de données de conduite autonome |
| Supervision cloud | Externe | Variable (fibre, 5G, VPN) | Temps réel, maintenance prédictive, vue globale du réseau | Dépend de la sécurité et de la disponibilité du réseau IP | Centre de contrôle recevant les données de plusieurs trains |

**Question de reflexion :** Pourquoi est-il nécessaire de migrer de **GSM-R** vers **FRMCS** ?  
Quels sont les avantages pour un TGV autonome ?  

👉 Écrire la réponse ci-dessous.

**Pourquoi migrer de GSM-R vers FRMCS ?**

GSM-R repose sur une technologie 2G obsolète (débit ~9,6 kbit/s), dont le support industriel est prévu jusqu'à 2030 seulement. Il ne permet pas de transmettre les gros volumes de données nécessaires aux applications modernes (vidéo, IA embarquée, supervision temps réel).

**Avantages du FRMCS pour un TGV autonome :**

- **Très haut débit et faible latence (5G)** : indispensable pour transmettre en temps réel les flux vidéo (caméras, LIDAR) et les décisions de l'IA embarquée entre le train et le centre de contrôle.
- **Support natif de l'IoT ferroviaire** : permet de connecter simultanément de nombreux capteurs (freins, portes, température, énergie) sans saturer le réseau, contrairement au GSM-R limité en débit.
- **Conduite autonome** : la 5G offre la réactivité nécessaire pour des échanges quasi instantanés entre le train, la signalisation et le centre de supervision — condition indispensable à toute automatisation poussée (ex. ETCS niveau 3).
- **Pérennité technologique** : succède à un système en fin de vie, garantissant l'interopérabilité européenne à long terme.
- **Redondance et fiabilité accrues** : architecture IP moderne facilitant la bascule automatique en cas de perte de liaison, un enjeu de sécurité critique.

## 🔹 Modules proposés

````python
import tkinter as tk
import random

root = tk.Tk()
root.title("Simulation de train")
root.geometry("600x400")

# Variables simulées
vitesse = tk.IntVar(value=0)
porte_fermee = tk.BooleanVar(value=True)
temperature = tk.IntVar(value=20)
signalisation = tk.StringVar(value="Voie libre")
lidar_obstacle = tk.BooleanVar(value=False)
gps_position = tk.StringVar(value="48.8566N, 2.3522E")
reseau_interne = tk.BooleanVar(value=True)
reseau_gsmr = tk.BooleanVar(value=True)
reseau_frmcs = tk.BooleanVar(value=True)
energie = tk.IntVar(value=50)
wifi = tk.BooleanVar(value=True)
etat_global = tk.StringVar(value="✅ OK – circulation normale")


def analyse_IA():
    if (not porte_fermee.get()) or (signalisation.get() == "Arrêt") or lidar_obstacle.get():
        # porte ouverte, OU signal à l'arrêt, OU obstacle détecté
        etat_global.set("🚨 Danger : départ interdit")
    elif not reseau_gsmr.get() and not reseau_frmcs.get():
        # GSM-R ET FRMCS coupés en même temps : plus aucune liaison sol-train disponible
        etat_global.set("🚨 Danger : perte totale de communication")
    elif vitesse.get() > 300 or temperature.get() > 40:
        etat_global.set("⚠️ Alerte")
    elif not reseau_interne.get():
        etat_global.set("⚠️ Alerte : réseau interne défaillant")
    elif energie.get() < 15:
        etat_global.set("⚠️ Alerte : énergie faible")
    else:
        etat_global.set("✅ OK – circulation normale")

    # Couleur de la barre selon la gravité de l'état
    texte = etat_global.get()
    if "Danger" in texte:              # vérifie si le mot "Danger" apparaît dans le texte
        label_etat.config(bg="#e74c3c")   # fond rouge
    elif "Alerte" in texte:
        label_etat.config(bg="#f39c12")   # fond orange
    else:
        label_etat.config(bg="#2ecc71")   # fond vert


def maj_donnees():
    vitesse.set(random.randint(0, 320))
    temperature.set(random.randint(15, 45))
    porte_fermee.set(random.choice([True, False]))
    signalisation.set(random.choice(["Voie libre", "Ralentissement", "Arrêt"]))
    lidar_obstacle.set(random.choice([True, False]))
    gps_position.set(f"{round(random.uniform(48,49),4)}N, {round(random.uniform(2,3),4)}E")
    reseau_interne.set(random.choice([True, False]))
    reseau_gsmr.set(random.choice([True, False]))
    reseau_frmcs.set(True)
    energie.set(random.randint(20, 100))
    wifi.set(random.choice([True, False]))

    analyse_IA()
    root.after(2000, maj_donnees)   # se reprogramme elle-même dans 2s, sans bloquer l'interface




# --- Interface ---
frame_top = tk.Frame(root, bg="lightblue")
frame_top.pack(fill="x")
tk.Label(frame_top, text="Vitesse (km/h)", bg="lightblue").pack()
tk.Label(frame_top, textvariable=vitesse, font=("Arial", 24), bg="lightblue").pack()

# Barre colorée pour l'état global (couleur changée dynamiquement dans analyse_IA)
label_etat = tk.Label(root, textvariable=etat_global, font=("Arial", 14, "bold"),
                       fg="white", bg="#2ecc71", pady=8)
label_etat.pack(fill="x")

# Chaque catégorie est séparée dans son propre bloc, avec un titre et une bordure
frame_securite = tk.Frame(root, bd=2, relief="groove")
frame_securite.pack(fill="x", padx=8, pady=4)
tk.Label(frame_securite, text="Sécurité", font=("Arial", 10, "bold")).pack(anchor="w", padx=6)
ligne_securite = tk.Frame(frame_securite)   # sous-cadre qui contient les infos
ligne_securite.pack(fill="x", padx=6, pady=(0, 6))
tk.Label(ligne_securite, text="Portes :").pack(side="left")
tk.Label(ligne_securite, textvariable=porte_fermee).pack(side="left")
tk.Label(ligne_securite, text="  Signalisation :").pack(side="left")
tk.Label(ligne_securite, textvariable=signalisation).pack(side="left")
tk.Label(ligne_securite, text="  LIDAR obstacle :").pack(side="left")
tk.Label(ligne_securite, textvariable=lidar_obstacle).pack(side="left")

frame_reseau = tk.Frame(root, bd=2, relief="groove")
frame_reseau.pack(fill="x", padx=8, pady=4)
tk.Label(frame_reseau, text="Réseau", font=("Arial", 10, "bold")).pack(anchor="w", padx=6)
ligne_reseau = tk.Frame(frame_reseau)
ligne_reseau.pack(fill="x", padx=6, pady=(0, 6))
tk.Label(ligne_reseau, text="Réseau interne :").pack(side="left")
tk.Label(ligne_reseau, textvariable=reseau_interne).pack(side="left")
tk.Label(ligne_reseau, text="  GSM-R :").pack(side="left")
tk.Label(ligne_reseau, textvariable=reseau_gsmr).pack(side="left")
tk.Label(ligne_reseau, text="  FRMCS :").pack(side="left")
tk.Label(ligne_reseau, textvariable=reseau_frmcs).pack(side="left")

frame_confort = tk.Frame(root, bd=2, relief="groove")
frame_confort.pack(fill="x", padx=8, pady=4)
tk.Label(frame_confort, text="Confort", font=("Arial", 10, "bold")).pack(anchor="w", padx=6)
ligne_confort = tk.Frame(frame_confort)
ligne_confort.pack(fill="x", padx=6, pady=(0, 6))
tk.Label(ligne_confort, text="Température :").pack(side="left")
tk.Label(ligne_confort, textvariable=temperature).pack(side="left")
tk.Label(ligne_confort, text="  Énergie :").pack(side="left")
tk.Label(ligne_confort, textvariable=energie).pack(side="left")
tk.Label(ligne_confort, text="  Wifi :").pack(side="left")
tk.Label(ligne_confort, textvariable=wifi).pack(side="left")

frame_gps = tk.Frame(root, bd=2, relief="groove")
frame_gps.pack(fill="x", padx=8, pady=4)
tk.Label(frame_gps, text="Position", font=("Arial", 10, "bold")).pack(anchor="w", padx=6)
ligne_gps = tk.Frame(frame_gps)
ligne_gps.pack(fill="x", padx=6, pady=(0, 6))
tk.Label(ligne_gps, text="GPS :").pack(side="left")
tk.Label(ligne_gps, textvariable=gps_position).pack(side="left")

maj_donnees()
root.mainloop()
````
